```python
print("---- SIMPLE PYTHON To-Do List ----")
tasks=[]
while True:
  print("1.Add Task")
  print("2.View Task")
  print("3.Delete Task")
  print("4.Exit")
  try:
    a=int(input("Enter your choice: "))
  except ValueError:
    print("Please enter a valid number\n")
    continue
  
  if a==1:
    task=input("Enter your task: ")
    tasks.append(task)
    print("Task Added Successfully \n")
  elif a==2:
    print("Your tasks: ")
    for i,task in enumerate(tasks,start=1):
      print(f"{i}. {task}")
    print("\n")
  elif a==3:
    if not tasks:
      print("No tasks availaible to delete\n")
    else:
      print("Your task: ")
      for i,task in enumerate(tasks,start=1):
       print(f"{i}. {task}")
      print("\n")
      while True:
        try:
          del_task=int(input("Enter the task to be deleted: "))
          if 1<=del_task<=len(tasks):
            tasks.pop(del_task-1)
            print("Successfully deleted the task\n")
            break
          else:
            print("Invalid Task Number\n")
        except ValueError:
          print("Enter Numbers,not the letters. Try again\n")

  elif a==4:
    print("\n Program Exiting..")
    break

  else:
    print("Invalid choice")





```
